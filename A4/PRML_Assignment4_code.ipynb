```python
import numpy as np
import math
import random
import matplotlib.pyplot as plt


# Step 1: Generate Gaussian variables using Assignment 1

def gaussian_pair():

    while True:

        u1 = 2 * random.random() - 1
        u2 = 2 * random.random() - 1

        s = u1**2 + u2**2

        if 0 < s < 1:

            # Assignment 1:
            # k = sqrt(-2 ln(s) / s)

            k = math.sqrt(-2 * math.log(s) / s)

            # x = u1*k
            # y = u2*k

            x = u1 * k
            y = u2 * k

            return x, y


# Step 2: Generate D1 and D2 with given means and variance

def generate_D1_D2(mu1, mu2, variance, n):

    D1 = []
    D2 = []

    sigma = math.sqrt(variance)

    for i in range(n):

        x, y = gaussian_pair()

        # Assignment 1:
        # x' = mu + sigma*x

        D1.append(mu1 + sigma * x)
        D2.append(mu2 + sigma * y)

    return np.array(D1), np.array(D2)


# Step 3: Form 2-D data

def form_data(D1, D2):

    return np.column_stack((D1, D2))


# Step 4: Matrix operations

def matrix_add(A, B):

    return np.add(A, B)


def matrix_multiply(A, B):

    return np.matmul(A, B)


def matrix_transpose(A):

    return np.transpose(A)


def matrix_inverse(A):

    return np.linalg.inv(A)


# Step 5: Calculate mean and covariance

def calculate_covariance(data):

    mean = np.mean(data, axis=0)

    centered = data - mean

    # Sigma = (1/N)(X-mu)^T(X-mu)

    covariance = matrix_multiply(
        matrix_transpose(centered),
        centered
    ) / len(data)

    return mean, covariance


# Step 6: Calculate eigenvalues and eigenvectors

def calculate_eigen(C):

    values, vectors = np.linalg.eigh(C)

    order = np.argsort(values)[::-1]

    values = values[order]
    vectors = vectors[:, order]

    return values, vectors


# Step 7: Calculate Sigma^(1/2)

def covariance_sqrt(C):

    values, vectors = calculate_eigen(C)

    # Sigma = Q Lambda Q^T
    # Sigma^(1/2) = Q Lambda^(1/2) Q^T

    D = np.diag(np.sqrt(values))

    C_half = matrix_multiply(
        matrix_multiply(vectors, D),
        matrix_transpose(vectors)
    )

    return C_half


# Step 8: Generate data for changed covariance

def transform_data(X, mu, C):

    C_half = covariance_sqrt(C)

    centered = X - mu

    # Assignment 4:
    # Y = mu + Sigma^(1/2)(X-mu)

    transformed = matrix_multiply(
        centered,
        matrix_transpose(C_half)
    )

    Y = transformed + mu

    return Y


# Step 9: Display results

def show_results(data, C):

    mean, covariance = calculate_covariance(data)

    values, vectors = calculate_eigen(covariance)

    print("\nEstimated Mean:")
    print(mean)

    print("\nEstimated Covariance Matrix:")
    print(covariance)

    print("\nEigenvalues:")
    print(values)

    print("\nEigenvectors:")
    print(vectors)

    major_axis = math.sqrt(values[0])
    minor_axis = math.sqrt(values[1])

    print("\nMajor Semi-Axis Length for c = 1:")
    print(major_axis)

    print("\nMinor Semi-Axis Length for c = 1:")
    print(minor_axis)

    print("\nGiven Covariance Matrix:")
    print(C)

    given_values, given_vectors = calculate_eigen(C)

    print("\nEigenvalues of Given Covariance:")
    print(given_values)

    print("\nEigenvectors of Given Covariance:")
    print(given_vectors)

    print("\nComparison:")

    print(
        "Largest estimated eigenvalue:",
        values[0]
    )

    print(
        "Major semi-axis length:",
        major_axis
    )

    print(
        "Smallest estimated eigenvalue:",
        values[1]
    )

    print(
        "Minor semi-axis length:",
        minor_axis
    )


# Step 10: Plot data

def plot_data(data, title):

    plt.figure(figsize=(7, 6))

    plt.scatter(
        data[:, 0],
        data[:, 1],
        s=10,
        alpha=0.5,
        label="Data"
    )

    plt.xlabel("X1")
    plt.ylabel("X2")
    plt.title(title)

    plt.legend()
    plt.grid(True)
    plt.axis("equal")

    plt.show()


# Step 11: Plot constant density curves

def plot_density_curves(data, mu, C, title):

    values, vectors = calculate_eigen(C)

    plt.figure(figsize=(7, 6))

    plt.scatter(
        data[:, 0],
        data[:, 1],
        s=8,
        alpha=0.25,
        label="Data"
    )

    theta = np.linspace(
        0,
        2 * np.pi,
        360
    )

    for c in [1, 2, 3]:

        # (x-mu)^T Sigma^(-1)(x-mu) = c
        #
        # Major semi-axis = sqrt(c*lambda1)
        # Minor semi-axis = sqrt(c*lambda2)

        major = math.sqrt(c * values[0])
        minor = math.sqrt(c * values[1])

        ellipse = np.array([
            major * np.cos(theta),
            minor * np.sin(theta)
        ])

        ellipse = matrix_multiply(
            vectors,
            ellipse
        )

        ellipse[0] = ellipse[0] + mu[0]
        ellipse[1] = ellipse[1] + mu[1]

        plt.plot(
            ellipse[0],
            ellipse[1],
            label="c = " + str(c)
        )

    # Eigenvectors give the directions of the axes

    v1 = vectors[:, 0]
    v2 = vectors[:, 1]

    major = math.sqrt(values[0])
    minor = math.sqrt(values[1])

    # Major axis

    plt.plot(
        [
            mu[0] - major * v1[0],
            mu[0] + major * v1[0]
        ],
        [
            mu[1] - major * v1[1],
            mu[1] + major * v1[1]
        ],
        color="red",
        linewidth=2,
        label="Major axis"
    )

    # Minor axis

    plt.plot(
        [
            mu[0] - minor * v2[0],
            mu[0] + minor * v2[0]
        ],
        [
            mu[1] - minor * v2[1],
            mu[1] + minor * v2[1]
        ],
        color="green",
        linewidth=2,
        label="Minor axis"
    )

    plt.scatter(
        mu[0],
        mu[1],
        color="black",
        s=50,
        label="Mean"
    )

    plt.xlabel("X1")
    plt.ylabel("X2")
    plt.title(title)

    plt.legend()
    plt.grid(True)
    plt.axis("equal")

    plt.show()


# Step 12: Number of samples

N = 1000


# Step 13: Problem 1 - Isotropic covariance

mu1 = np.array([
    2.0,
    3.0
])

C1 = np.array([
    [1.0, 0.0],
    [0.0, 1.0]
])

D1, D2 = generate_D1_D2(
    mu1[0],
    mu1[1],
    1.0,
    N
)

X1 = form_data(
    D1,
    D2
)

D1_final = transform_data(
    X1,
    mu1,
    C1
)

print("PROBLEM 1: ISOTROPIC COVARIANCE")

print("\nGiven Mean:")
print(mu1)

print("\nGiven Covariance:")
print(C1)

show_results(
    D1_final,
    C1
)

plot_data(
    D1_final,
    "Problem 1: Isotropic Gaussian Data"
)

plot_density_curves(
    D1_final,
    mu1,
    C1,
    "Problem 1: Isotropic Constant Density Curves"
)


# Step 14: Problem 2 - Diagonal covariance

mu2 = np.array([
    2.0,
    3.0
])

C2 = np.array([
    [1.0, 0.0],
    [0.0, 4.0]
])

D1, D2 = generate_D1_D2(
    mu2[0],
    mu2[1],
    1.0,
    N
)

X2 = form_data(
    D1,
    D2
)

D2_final = transform_data(
    X2,
    mu2,
    C2
)

print("PROBLEM 2: DIAGONAL COVARIANCE")

print("\nGiven Mean:")
print(mu2)

print("\nGiven Covariance:")
print(C2)

show_results(
    D2_final,
    C2
)

plot_data(
    D2_final,
    "Problem 2: Diagonal Gaussian Data"
)

plot_density_curves(
    D2_final,
    mu2,
    C2,
    "Problem 2: Diagonal Constant Density Curves"
)


# Step 15: Problem 3 - Full covariance

mu3 = np.array([
    2.0,
    3.0
])

C3 = np.array([
    [4.0, 1.5],
    [1.5, 1.0]
])

D1, D2 = generate_D1_D2(
    mu3[0],
    mu3[1],
    1.0,
    N
)

X3 = form_data(
    D1,
    D2
)

D3_final = transform_data(
    X3,
    mu3,
    C3
)

print("PROBLEM 3: FULL COVARIANCE")

print("\nGiven Mean:")
print(mu3)

print("\nGiven Covariance:")
print(C3)

show_results(
    D3_final,
    C3
)

plot_data(
    D3_final,
    "Problem 3: Full Covariance Gaussian Data"
)

plot_density_curves(
    D3_final,
    mu3,
    C3,
    "Problem 3: Full Covariance Constant Density Curves"
)


# Step 16: Problem 4 - Generate two Gaussian classes

def generate_class(mu, C, n):

    D1, D2 = generate_D1_D2(
        mu[0],
        mu[1],
        1.0,
        n
    )

    X = form_data(
        D1,
        D2
    )

    Y = transform_data(
        X,
        mu,
        C
    )

    return Y


# Step 17: Plot two classes

def plot_two_classes(
    Y1,
    Y2,
    mu1,
    mu2,
    title,
    boundary=False
):

    plt.figure(figsize=(8, 7))

    plt.scatter(
        Y1[:, 0],
        Y1[:, 1],
        s=10,
        alpha=0.5,
        label="Class omega1"
    )

    plt.scatter(
        Y2[:, 0],
        Y2[:, 1],
        s=10,
        alpha=0.5,
        label="Class omega2"
    )

    plt.scatter(
        mu1[0],
        mu1[1],
        color="black",
        s=60
    )

    plt.scatter(
        mu2[0],
        mu2[1],
        color="black",
        s=60
    )

    if boundary:

        # For equal covariance and equal priors:
        # ||x-mu1||^2 = ||x-mu2||^2
        #
        # mu1 = (0,0)
        # mu2 = (5,5)
        #
        # x1 + x2 = 5

        x = np.linspace(-4, 9, 200)

        y = 5 - x

        plt.plot(
            x,
            y,
            linewidth=2,
            label="Decision Boundary"
        )

    plt.xlabel("X1")
    plt.ylabel("X2")
    plt.title(title)

    plt.legend()
    plt.grid(True)
    plt.axis("equal")

    plt.show()


# Step 18: Problem 4 - Experiment 1

mu_class1 = np.array([
    0.0,
    0.0
])

mu_class2 = np.array([
    5.0,
    5.0
])

C_class1 = np.array([
    [1.0, 0.0],
    [0.0, 1.0]
])

C_class2 = np.array([
    [1.0, 0.0],
    [0.0, 1.0]
])

Y1 = generate_class(
    mu_class1,
    C_class1,
    N
)

Y2 = generate_class(
    mu_class2,
    C_class2,
    N
)

print("PROBLEM 4: EX 1")

print("\nClass omega1")

show_results(
    Y1,
    C_class1
)

print("\nClass omega2")

show_results(
    Y2,
    C_class2
)

plot_two_classes(
    Y1,
    Y2,
    mu_class1,
    mu_class2,
    "Problem 4: Ex 1",
    True
)


# Step 19: Problem 4 - Ex 2

mu_class1 = np.array([
    0.0,
    0.0
])

mu_class2 = np.array([
    5.0,
    5.0
])

C_class1 = np.array([
    [1.0, 0.0],
    [0.0, 1.0]
])

C_class2 = np.array([
    [4.0, 0.0],
    [0.0, 1.0]
])

Y1 = generate_class(
    mu_class1,
    C_class1,
    N
)

Y2 = generate_class(
    mu_class2,
    C_class2,
    N
)

print("PROBLEM 4: EX 2")

print("\nClass omega1")

show_results(
    Y1,
    C_class1
)

print("\nClass omega2")

show_results(
    Y2,
    C_class2
)

plot_two_classes(
    Y1,
    Y2,
    mu_class1,
    mu_class2,
    "Problem 4: EX 2",
    False
)


# Step 20: Problem 4 - Ex 3

mu_class1 = np.array([
    0.0,
    0.0
])

mu_class2 = np.array([
    5.0,
    5.0
])

C_class1 = np.array([
    [2.0, 1.0],
    [1.0, 2.0]
])

C_class2 = np.array([
    [2.0, -1.0],
    [-1.0, 2.0]
])

Y1 = generate_class(
    mu_class1,
    C_class1,
    N
)

Y2 = generate_class(
    mu_class2,
    C_class2,
    N
)

print("PROBLEM 4: EXPERIMENT 3")

print("\nClass omega1")

show_results(
    Y1,
    C_class1
)

print("\nClass omega2")

show_results(
    Y2,
    C_class2
)

plot_two_classes(
    Y1,
    Y2,
    mu_class1,
    mu_class2,
    "Problem 4: Ex 3",
    False
)

```
